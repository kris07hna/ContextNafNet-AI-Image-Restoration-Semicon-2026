```python
# Step 4: Setup Input Directory for Inference
import os, numpy as np
os.makedirs('demo_input', exist_ok=True)
if not os.path.exists('Test_NoisyLR/NoisyLR'):
    demo_img = np.random.rand(128, 128).astype(np.float32)
    np.save('demo_input/demo_001.npy', demo_img)
    input_dir = 'demo_input'
else:
    input_dir = 'Test_NoisyLR/NoisyLR'
print(f"Using input directory: {input_dir}")
```

```python
# Step 5: Execute KLA Wafer Restoration Submission Command
!python run.py {input_dir} colab_outputs
```

```python
# Step 6: Visualize Original Noisy Input vs Restored Output
import matplotlib.pyplot as plt, numpy as np, glob
out_files = sorted(glob.glob('colab_outputs/*.npy'))
if out_files:
    sample_out = out_files[0]
    sample_name = os.path.basename(sample_out)
    sample_in = os.path.join(input_dir, sample_name)
    noisy = np.load(sample_in) if os.path.exists(sample_in) else np.zeros((128, 128))
    restored = np.load(sample_out)
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(noisy, cmap='gray'); axes[0].set_title(f'Noisy Input ({noisy.shape})'); axes[0].axis('off')
    axes[1].imshow(restored, cmap='gray'); axes[1].set_title(f'Restored Output ({restored.shape})'); axes[1].axis('off')
    plt.tight_layout(); plt.show()
    print(f"Successfully restored wafer saved with shape {restored.shape}, min={restored.min():.4f}, max={restored.max():.4f}")
```
